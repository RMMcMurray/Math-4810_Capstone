```python
#| label: melanie setup
#| include: false

def normalize(df):
    """
    Returns two dataframes:
    1. df_quantile: Everything forced to normal via Quantile Transformation.
    2. df_comp: Each column transformed by the 'winner' of YJ vs. Quantile.
    """
    df_quantile = pd.DataFrame(index=df.index)
    df_comp = pd.DataFrame(index=df.index)
    
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    
    for col in numeric_cols:
        data_raw = df[col].values.reshape(-1, 1)

        # Yeo-Johnson
        yj_data, lmbda = stats.yeojohnson(df[col])
        _, yj_p = stats.normaltest(yj_data)
        
        # Normality test for the Quantile data
        qt_engine = QuantileTransformer(output_distribution='normal', 
                                        n_quantiles=min(len(df), 1000), 
                                        random_state=42)
        qt_data = qt_engine.fit_transform(data_raw).flatten()
        _, qt_p = stats.normaltest(qt_data)
        
        if yj_p >= qt_p:
            # Determine suffix for YJ winner
            if -0.1 <= lmbda <= 0.1: suffix = "log"
            elif 0.4 <= lmbda <= 0.6: suffix = "sqrt"
            elif 0.9 <= lmbda <= 1.1: suffix = "linear"
            elif lmbda < 0: suffix = "inv"
            else: suffix = "pow"
            df_comp[f"{col}_{suffix}"] = yj_data
        else:
            df_comp[f"{col}_quant"] = qt_data
            
    print(f"Experiments complete for {len(numeric_cols)} variables.")
    return df_comp

parcel_export = pd.read_excel("/Users/melaniejensen/Library/CloudStorage/OneDrive-SouthernUtahUniversity/Math-4810_Capstone/data/Parcel Export.xlsx", sheet_name="Full Parcel Summary")
parcel_export.head()
parcel_export.shape # (53277, 53)
parcel_export.info()

non_numeric_cols = parcel_export.select_dtypes(exclude=[np.number]).columns.tolist()
parcel_export[non_numeric_cols].shape # (53277, 12)
parcel_export[non_numeric_cols].info()
parcel_export[non_numeric_cols].columns

numeric_cols = parcel_export.select_dtypes(include=[np.number]).columns.tolist() 
parcel_export[numeric_cols].shape # (53277, 41)
parcel_export[numeric_cols].info() 
parcel_export[numeric_cols].columns

cols_to_move = ['TaxYear', 'ImpOnly', 'PropType ', 'Jurisdiction', 'SpecificPropType', 'Lat', 'Lng', 'ReviewedDate', 'ImpCode', 'Res Imp Count']
non_numeric_cols = list(set(non_numeric_cols) | set(cols_to_move))
numeric_cols = [c for c in parcel_export.columns if c not in non_numeric_cols]

parcel_export[numeric_cols].shape # (53277, 31)

parcel_export[numeric_cols] = parcel_export[numeric_cols].fillna(0)

numeric_parcel_data = parcel_export[numeric_cols]

numeric_parcel_data.shape # (53277, 24)
numeric_parcel_data.info()

df_competition = normalize(numeric_parcel_data)
df_competition.columns

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_competition)

pca_full = PCA(n_components=None)
pca_full.fit(X_scaled)
```

```python
#| label: ryan-setup
#| include: false

res = pd.read_excel(
    "/Users/melaniejensen/Library/CloudStorage/OneDrive-SouthernUtahUniversity/Math-4810_Capstone/data/2025 MLS Residential Sales.xlsm", sheet_name="Sales").assign(segment="Residential")

multi = pd.read_excel(
    "/Users/melaniejensen/Library/CloudStorage/OneDrive-SouthernUtahUniversity/Math-4810_Capstone/data/2022 - 2025 Multi-Family MLS Export.xlsm", sheet_name="textexport (3)").assign(segment="Multi-Family")

land = pd.read_excel(
    "/Users/melaniejensen/Library/CloudStorage/OneDrive-SouthernUtahUniversity/Math-4810_Capstone/data/2022 - 2025 MLS Land Sales.xlsm", sheet_name="Sales").assign(segment="Land Sales")

# Stack and Merge
all_sales = pd.concat([res, multi, land], ignore_index=True)
parcels = parcel_export

df_combined = pd.merge(
    all_sales, parcels, left_on="Parcel Number", right_on="ParcelId", how="left"
)
df_combined["Sold Date"] = pd.to_datetime(df_combined["Sold Date"])

# Create res_df for the segmentation section later
res_df = df_combined[df_combined["segment"] == "Residential"].copy()

df_combined["Adj_Price"] = df_combined["Sold Price"] - df_combined[
    "Concessions Amount"
].fillna(0)
df_combined = df_combined[df_combined["Adj_Price"] > 0].copy()
df_combined["ln_Price"] = np.log(df_combined["Adj_Price"])
df_combined["ln_GLA"] = np.log(df_combined["Total SqFt"].replace(0, np.nan))

start_date = pd.Timestamp("2021-01-01")
lien_date = pd.Timestamp("2026-01-01")
x_days = (df_combined["Sold Date"] - start_date).dt.days.values

pw_model = pwlf.PiecewiseLinFit(x_days, df_combined["ln_Price"].values)
knots_days = pw_model.fit(3)
knot_dates = [start_date + pd.Timedelta(days=int(k)) for k in knots_days]

for i in range(len(knot_dates) - 1):
    col_name = f"Time_{i + 1}"
    start, end = knot_dates[i], knot_dates[i + 1]
    df_combined[col_name] = df_combined["Sold Date"].apply(
        lambda x: max(0, (min(x, end) - start).days / 30.44) if pd.notna(x) else 0
    )

df_combined["Trended_Price"] = df_combined["Adj_Price"].copy()
time_cols = [c for c in df_combined.columns if c.startswith("Time_")]

for seg in ["Residential", "Multi-Family", "Land Sales"]:
    df_seg = df_combined[df_combined["segment"] == seg].copy()
    if len(df_seg) < 20:
        continue
    valid_buckets = [col for col in time_cols if df_seg[col].max() > 0]
    formula = (
        f"ln_Price ~ {' + '.join(valid_buckets)} + ln_GLA + Q('Year Built')"
        if seg != "Land Sales"
        else f"ln_Price ~ {' + '.join(valid_buckets)}"
    )
    try:
        res_ols = smf.ols(
            formula,
            data=df_seg.dropna(
                subset=["ln_GLA", "Year Built"] if seg != "Land Sales" else []
            ),
        ).fit()
        # Trending Logic
        adj_log = 0
        for idx, col in enumerate(time_cols):
            if col in res_ols.params:
                remaining = (
                    (knot_dates[idx + 1] - knot_dates[idx]).days / 30.44
                ) - df_combined.loc[df_combined["segment"] == seg, col]
                adj_log += res_ols.params[col] * np.maximum(0, remaining)
        df_combined.loc[df_combined["segment"] == seg, "Trended_Price"] *= np.exp(
            adj_log
        )
    except:
        pass

# Data Preparation
df_plot = df_combined[
    (df_combined["segment"] == "Residential") & (df_combined["Total SqFt"] > 100)
].copy()

# Calculate Unit Prices
df_plot["Price_SqFt_Original"] = df_plot["Adj_Price"] / df_plot["Total SqFt"]
df_plot["Price_SqFt_Trended"] = df_plot["Trended_Price"] / df_plot["Total SqFt"]

# Outlier Cleaning
y_limit = df_plot["Price_SqFt_Original"].quantile(0.98)
df_plot = df_plot[df_plot["Price_SqFt_Original"] < y_limit]

# Numeric conversion for regression plotting
df_plot["date_num"] = mdates.date2num(df_plot["Sold Date"])

# Plotting
plt.figure(figsize=(16, 8))

# Original Market
sns.regplot(
    data=df_plot,
    x="date_num",
    y="Price_SqFt_Original",
    scatter_kws={"alpha": 0.1, "color": "gray", "s": 12},
    line_kws={
        "color": "#e74c3c",
        "label": "Original Market (Volatile)",
        "linewidth": 3,
    },
    lowess=True,
)

# Trended Market
sns.regplot(
    data=df_plot,
    x="date_num",
    y="Price_SqFt_Trended",
    scatter=False,
    line_kws={
        "color": "#3498db",
        "label": f"Trended to {lien_date.date()} (Stabilized)",
        "linewidth": 4,
    },
    lowess=False,
)

# The Splines
for s in knot_dates:
    plt.axvline(
        mdates.date2num(s), color="black", linestyle="--", alpha=0.2, label="_nolegend_"
    )

# Formatting for Revealjs
ax = plt.gca()
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))
plt.ylim(0, y_limit)
plt.title("Residential Housing Market Stabilization", fontsize=22, pad=20)
plt.ylabel("Price per Square Foot ($)", fontsize=14)
plt.xlabel("Sale Date", fontsize=14)
plt.legend(
    loc="upper left", fontsize=13, frameon=True, facecolor="white", framealpha=0.9
)
plt.grid(axis="y", alpha=0.15)

plt.tight_layout()
plt.show()


# --- CORE UTILITIES ---
def get_scaled_data(df):
    # Strictly isolate numbers BEFORE doing any math
    X = df.select_dtypes(include=[np.number]).fillna(0)
    # Remove columns with zero variance (they break PCA)
    X = X.loc[:, X.var() > 1e-4]
    return StandardScaler().fit_transform(X)


def get_umap_embedding(scaled_data, nn, md):
    # Reduce dimensions via PCA first for UMAP stability
    pca_data = PCA(n_components=0.9).fit_transform(scaled_data)
    reducer = umap.UMAP(n_neighbors=nn, min_dist=md, random_state=42)
    return reducer.fit_transform(pca_data)


def get_clusters(pca_data, min_size=20):
    clusterer = hdbscan.HDBSCAN(min_cluster_size=min_size, prediction_data=True)
    return clusterer.fit_predict(pca_data)


# --- VISUALIZATION WRAPPERS ---
def plot_umap_branded(embedding, color_source, title, is_price=True):
    plt.figure(figsize=(10, 6))
    if is_price:
        price_ranks = rankdata(color_source) / len(color_source)
        scatter = plt.scatter(
            embedding[:, 0],
            embedding[:, 1],
            c=price_ranks,
            cmap="viridis",
            s=10,
            alpha=0.6,
        )
        plt.colorbar(scatter).set_label("Price Percentile (0=Low, 1=High)")
    else:
        unique_labels = np.unique(color_source)
        cmap = plt.get_cmap("tab10", len(unique_labels))
        for i, label in enumerate(unique_labels):
            mask = color_source == label
            plt.scatter(
                embedding[mask, 0],
                embedding[mask, 1],
                label=f"Cluster {label}" if label != -1 else "Noise",
                s=15,
                alpha=0.8,
                color=cmap(i),
            )
        plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.title(title)
    plt.tight_layout()
    plt.show()


def plot_pca_heatmap(pca_model, df_features):
    # FIX: Isolate numeric columns specifically to avoid String -> Float errors
    numeric_df = df_features.select_dtypes(include=[np.number]).fillna(0)
    kept_columns = numeric_df.columns[numeric_df.var() > 1e-4]

    # Create the loadings dataframe
    loadings_df = pd.DataFrame(pca_model.components_.T, index=kept_columns)

    plt.figure(figsize=(12, 8))
    # Show only top 15 features to keep the heatmap readable
    sns.heatmap(loadings_df.head(15), annot=True, cmap="RdBu_r", center=0, fmt=".2f")
    plt.title("PCA Loadings: Feature Correlation to Components")
    plt.show()


# --- EXECUTION ---
# Ensure res_df is defined from the main df_combined
res_df = df_combined[df_combined["segment"] == "Residential"].copy()

# Drop columns that shouldn't be used for clustering/PCA
res_features = res_df.drop(
    columns=["Sold Price", "Adj_Price", "Trended_Price", "List Number", "ParcelId"],
    errors="ignore",
)

res_scaled = get_scaled_data(res_features)
res_pca_model = PCA(n_components=0.9).fit(res_scaled)
res_pca_data = res_pca_model.transform(res_scaled)
res_emb_l = get_umap_embedding(res_scaled, nn=15, md=0.01)
res_l_clusters = get_clusters(res_pca_data, min_size=30)
res_target = res_df["Trended_Price"]
```

```python
#| label: unified_price_adjustment
#| warning: false

df_combined["Adj_Price"] = df_combined["Sold Price"] - df_combined[
    "Concessions Amount"
].fillna(0)
df_combined = df_combined[df_combined["Adj_Price"] > 0].copy()
df_combined["ln_Price"] = np.log(df_combined["Adj_Price"])
df_combined["ln_GLA"] = np.log(df_combined["Total SqFt"].replace(0, np.nan))

start_date = pd.Timestamp("2021-01-01")
lien_date = pd.Timestamp("2026-01-01")
x_days = (df_combined["Sold Date"] - start_date).dt.days.values

pw_model = pwlf.PiecewiseLinFit(x_days, df_combined["ln_Price"].values)
knots_days = pw_model.fit(3)
knot_dates = [start_date + pd.Timedelta(days=int(k)) for k in knots_days]

for i in range(len(knot_dates) - 1):
    col_name = f"Time_{i + 1}"
    start, end = knot_dates[i], knot_dates[i + 1]
    df_combined[col_name] = df_combined["Sold Date"].apply(
        lambda x: max(0, (min(x, end) - start).days / 30.44) if pd.notna(x) else 0
    )

df_combined["Trended_Price"] = df_combined["Adj_Price"].copy()
time_cols = [c for c in df_combined.columns if c.startswith("Time_")]

for seg in ["Residential", "Multi-Family", "Land Sales"]:
    df_seg = df_combined[df_combined["segment"] == seg].copy()
    if len(df_seg) < 20:
        continue
    valid_buckets = [col for col in time_cols if df_seg[col].max() > 0]
    formula = (
        f"ln_Price ~ {' + '.join(valid_buckets)} + ln_GLA + Q('Year Built')"
        if seg != "Land Sales"
        else f"ln_Price ~ {' + '.join(valid_buckets)}"
    )
    try:
        res_ols = smf.ols(
            formula,
            data=df_seg.dropna(
                subset=["ln_GLA", "Year Built"] if seg != "Land Sales" else []
            ),
        ).fit()
        # Trending Logic
        adj_log = 0
        for idx, col in enumerate(time_cols):
            if col in res_ols.params:
                remaining = (
                    (knot_dates[idx + 1] - knot_dates[idx]).days / 30.44
                ) - df_combined.loc[df_combined["segment"] == seg, col]
                adj_log += res_ols.params[col] * np.maximum(0, remaining)
        df_combined.loc[df_combined["segment"] == seg, "Trended_Price"] *= np.exp(
            adj_log
        )
    except:
        pass
```
